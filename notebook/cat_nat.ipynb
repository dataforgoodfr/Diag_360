```python
import duckdb
import pandas as pd
```

```python
df_epci = pd.read_csv("../data/processed/epci_membres.csv")
```

```python
query = """
SELECT 
    siren,
    sum(superficie_km2) AS superficie_km2
FROM df_epci
GROUP BY siren
"""

df_surface_epci = duckdb.sql(query)
df_surface_epci
```

```python
df_cat_nat = pd.read_csv("../data/data_cat_nat/raw/catnat_gaspar.csv", sep=";")
df_cat_nat
```

```python
query = """ 
SELECT 
    count(dat_pub_arrete) AS nb_cat_nat,
    siren,
    epci_nom,
    dept_epci
FROM df_epci
LEFT JOIN df_cat_nat 
ON df_epci.code_insee = df_cat_nat.cod_commune
GROUP BY siren, epci_nom, dept_epci
ORDER BY siren
"""

duckdb.sql(query).df()
```

```python
query = """
SELECT e.siren,
        e.dept,
        e.total_pop_tot,
    df_cat_nat.nb_cat_nat,
    ROUND(1.0*df_cat_nat.nb_cat_nat / e.superficie_km2, 3) AS cat_nat_per_km2
FROM df_epci AS e
LEFT JOIN df_cat_nat
ON e.siren = df_cat_nat.siren
"""

df_cat_nat_final = duckdb.sql(query)
```

```python
df_cat_nat_final.write_csv("./data/processed/cat_nat_epci.csv")
```
