```python
import duckdb
import pandas as pd
import requests
```

```python
url_json = (
    "https://www.data.gouv.fr/api/1/datasets/r/4f02ce39-2a91-4a8b-85cb-c6a0f912516b"
)
response = requests.get(url_json)
data = response.json()
```

```python
df = pd.json_normalize(
    data,
    record_path="membres",
    meta=[
        "code",
        "nom",
        "type",
        "modeFinancement",
        "populationTotale",
        "populationMunicipale",
    ],
    record_prefix="",
    meta_prefix="epci_",
)
```
